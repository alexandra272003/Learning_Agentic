## Zero Shot Chain Of Thoughts

```python

from dotenv import load_dotenv
from openai import OpenAI
import os

load_dotenv()

client = OpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)

# Zero-shot CoT suffix
zero_shot_suffix = "Let's think step by step."

# Create zero-shot CoT prompt
def zero_shot_cot_prompt(question: str) -> str:
    return f"{question} {zero_shot_suffix}"

# Question
question = "X has five balls. He bought two more cans of 3 balls. How many balls does he have?"

# Create prompt
prompt = zero_shot_cot_prompt(question)

# Call Groq LLM
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": prompt}]
)

# Get the reply
llm_reply_zero_shot = response.choices[0].message.content

print(llm_reply_zero_shot)
```

## Few Shot Chain Of Thoughts

```python
# Two examples
examples = """
Q: X has 2 balls and buys 3 more balls.
A: 2 + 3 = 5. Answer: 5 balls.

Q: X has 4 balls and buys 2 cans of 3 balls each.
A: 4 + (2 * 3) = 10. Answer: 10 balls.
"""

# Few-shot prompt
few_shot_prompt = f"""
{examples}

Q: X has five balls. He bought two cans of 3 balls.
How many balls does he have?

Let's think step by step.
"""

# Call Groq LLM
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": few_shot_prompt}]
)

llm_reply_few_shot = response.choices[0].message.content
print(llm_reply_few_shot)
```
